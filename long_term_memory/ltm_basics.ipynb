```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
namespace = ("user", "ul")
```

## Creating Memory

```python
store.put(namespace, "1", {"data", "User likes pizza"})
store.put(namespace, "2", {"data", "User prefers dark mode"})
```

```python
namespace2 = ("user", "u2")
store.put(namespace2, "1", {"data", "User likes pasta"})
store.put(namespace2, "2", {"data", "User prefers grid style navigation"})
```

## Retreiving Memories

```python
store.get(namespace2, "2")
```

## Retreiving all Memories

```python
items = store.search(namespace)
for item in items:
    print(item.value)
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
import os
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key=os.getenv("GOOGLE_GENAI_KEY")
)
```

```python
store = InMemoryStore(
    index = {
        "embed": embedding_model, "dim": 1536
    }
)
```

```python
namespace = ("users", "u1")
```

```python
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})
store.put(namespace, "3", {"data": "User lives in Chennai"})
store.put(namespace, "4", {"data": "User enjoys playing football"})
store.put(namespace, "5", {"data": "User prefers Python over Java"})
store.put(namespace, "6", {"data": "User is learning LangGraph"})
store.put(namespace, "7", {"data": "User likes coffee"})
store.put(namespace, "8", {"data": "User prefers Windows over macOS"})
store.put(namespace, "9", {"data": "User enjoys watching sci-fi movies"})
store.put(namespace, "10", {"data": "User prefers working at night"})
```

```python
items = store.search(
    namespace , query = "what is  user currently learning", limit =1
)

for item in items:
    print(item.value)
```

```python
items = store.search(
    namespace , query = "What are the users preferences", limit =3
)

for item in items:
    print(item.value)
```
