```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    google_api_key=os.getenv("GOOGLE_GENAI_KEY")
)
```

```python
class BlogState(TypedDict):
    topic: str
    outline: str
    blog_post: str
```

```python
def create_outline(state: BlogState) -> BlogState:
    topic = state['topic']
    #Outline prompt
    outline_prompt = f"Create a detailed outline for a blog post on the topic: {topic}"
    #Call llm
    outline = llm.invoke(outline_prompt).content[0]['text']
    #Update state
    state['outline'] = outline
    return state

def write_blog_post(state: BlogState) -> BlogState:
    outline = state['outline']
    topic = state['topic']
    #Create the prompt
    blog_prompt = f"Write a detailed blog post on the topic: {topic} based on the following outline: {outline}"
    #Call llm
    blog_post = llm.invoke(blog_prompt).content[0]['text']
    #Update state
    state['blog_post'] = blog_post
    return state
```

```python
#Create Graph
graph = StateGraph(BlogState)

#Add nodes
graph.add_node('create_outline', create_outline)
graph.add_node('write_blog_post', write_blog_post)

#Add edges
graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'write_blog_post')
graph.add_edge('write_blog_post', END)

#compile the graph into a workflow
workflow = graph.compile()
```

```python
initial_state = {
    'topic': 'The impact of AI on modern education',
    'outline': '',
    'blog_post': ''
}

final_state = workflow.invoke(initial_state)

print("Final State:", final_state['blog_post'])
```
