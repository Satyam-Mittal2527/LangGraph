```python
from langgraph.graph import StateGraph, START, END
from  typing import TypedDict, Literal
```

```python
class QuadState(TypedDict):
    a : int
    b: int
    c: int
    equation : str
    descriminant : float
    result : str
```

```python
def show_equation(state : QuadState) -> dict: 

    e = f'{state['a']}x2{state['b']}x{state['c']}'

    return {
        'equation': e
    }

def calculate_descrimnant(state: QuadState) -> dict:
    descrimnant = state['b']**2 - (4 * state['a'] * state['c'])

    return {
        'descriminant': descrimnant
    }

def real_roots(state : QuadState):

    root1 = (-state['b'] + state['descriminant']**0.5)/ (2 * state['a'])
    root2 = (-state['b'] - state['descriminant']**0.5) /(2 * state['a'])

    return{
        'result': f'The roots are {root1} and {root2}'
    }
def repeated_roots(state : QuadState):

    root = (-state['b']) / (2 * state['a'])
    
    return{
        'result': f'Only repating root is {root}'
    }
def no_real_roots(state : QuadState):

    return{
        'result': f'No real roots possible'
    }
def check_condition(state: QuadState) -> Literal['real_roots', 'repeated_roots', 'no_real_roots']:
    if state['descriminant']> 0:
        return 'real_roots'
    elif state['descriminant']== 0:
        return 'repeated_roots'
    else: 
        return 'no_real_roots'
```

```python
#Make graph
graph = StateGraph(QuadState)

#Add nodes
graph.add_node('show_equation', show_equation)
graph.add_node('calculate_descrimnant', calculate_descrimnant)
graph.add_node('real_roots', real_roots)
graph.add_node('repeated_roots', repeated_roots)
graph.add_node('no_real_roots', no_real_roots)
#Add edge
graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation', 'calculate_descrimnant')
graph.add_conditional_edges('calculate_descrimnant', check_condition)
graph.add_edge('real_roots', END)
graph.add_edge('repeated_roots', END)
graph.add_edge('no_real_roots', END)


workflow = graph.compile()
```

```python
intial_state = {
    'a': 4,
    'b' : 2, 
    'c': 4
}

final_state = workflow.invoke(intial_state)

final_state
```
