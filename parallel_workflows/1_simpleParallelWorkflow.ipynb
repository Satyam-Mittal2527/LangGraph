```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
 
```

```python
class BatsManSat(TypedDict):
    runs: int
    balls: int
    fours_no: int
    sixes_no: int
    strike_rate: float
    boundary_percentage: float
    ballsPerBoundary: float
    summary: str
    
```

```python
def calculate_strike(state : BatsManSat) :
    sr =  (state['runs'] / state['balls']) * 100
    return {'strike_rate': sr}
def calculate_boundary(state : BatsManSat):
    bp = ((state['fours_no'] + state['sixes_no']) / state['balls']) * 100
    return {'boundary_percentage': bp}
def calculate_balls_per_boundary(state : BatsManSat):
    bpb = state['balls'] / (state['fours_no'] + state['sixes_no'])
    return {'ballsPerBoundary': bpb}    
def summary(state : BatsManSat):
    summary_text = f"Runs: {state['runs']}, Balls: {state['balls']}, Fours: {state['fours_no']}, Sixes: {state['sixes_no']}, Strike Rate: {state['strike_rate']:.2f}, Boundary Percentage: {state['boundary_percentage']:.2f}, Balls per Boundary: {state['ballsPerBoundary']:.2f}"
    return {'summary': summary_text}
```

```python
graph = StateGraph(BatsManSat)

#Add Nodes
graph.add_node('calculate_strike_rate', calculate_strike)
graph.add_node('calculate_boundary_percentage', calculate_boundary)
graph.add_node('calculate_balls_per_boundary', calculate_balls_per_boundary)
graph.add_node('generate_summary', summary)

#Add Edges
graph.add_edge(START, 'calculate_strike_rate')
graph.add_edge(START, 'calculate_boundary_percentage')
graph.add_edge(START, 'calculate_balls_per_boundary')
graph.add_edge('calculate_strike_rate', 'generate_summary')
graph.add_edge('calculate_boundary_percentage', 'generate_summary')
graph.add_edge('calculate_balls_per_boundary', 'generate_summary') 
graph.add_edge('generate_summary', END)

#Compile the graph
workflow = graph.compile()
```

```python
initial_state = {
    'runs': 100,
    'balls': 60,
    'fours_no': 10,
    'sixes_no': 5,
    'strike_rate': 0.0,
    'boundary_percentage': 0.0,
    'ballsPerBoundary': 0.0,
    'summary': ''
}

final_state = workflow.invoke(initial_state)

final_state
```
